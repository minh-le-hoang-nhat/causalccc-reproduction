Matrice obtenue avec kallisto|bustools (`run_kb.sh`). Adapté du tutoriel scanpy pbmc3k.

```python
from __future__ import annotations

import matplotlib.pyplot as plt
import pandas as pd
import scanpy as sc
import anndata as ad
from scipy.io import mmread
from scipy.sparse import csr_matrix
```

```python
sc.settings.verbosity = 3
sc.set_figure_params(dpi=80, facecolor="white")
sc.logging.print_header()
```

```python
SRR = "SRR11396171"
kb_dir = f"kb_out/{SRR}/counts_filtered"
results_file = f"write/{SRR}.h5ad"
```

```python
# matrice kb : cellules x genes
X = csr_matrix(mmread(f"{kb_dir}/cells_x_genes.mtx")).astype("float32")
barcodes = pd.read_csv(f"{kb_dir}/cells_x_genes.barcodes.txt", header=None)[0].values
gene_ids = pd.read_csv(f"{kb_dir}/cells_x_genes.genes.txt", header=None)[0].values
gene_names = pd.read_csv(f"{kb_dir}/cells_x_genes.genes.names.txt", header=None)[0].fillna(pd.Series(gene_ids)).values
adata = ad.AnnData(X=X, obs=pd.DataFrame(index=barcodes), var=pd.DataFrame({"gene_ids": gene_ids}, index=gene_names))
adata.var_names_make_unique()
adata.obs["sample"] = SRR
adata
```

```python
sc.pl.highest_expr_genes(adata, n_top=20)
```

```python
sc.pp.filter_cells(adata, min_genes=200)  # this does nothing, in this specific case
sc.pp.filter_genes(adata, min_cells=3)
adata
```

```python
# annotate the group of mitochondrial genes as "mt"
adata.var["mt"] = adata.var_names.str.startswith("MT-")
sc.pp.calculate_qc_metrics(adata, qc_vars=["mt"], percent_top=None, log1p=False, inplace=True)
```

```python
sc.pl.violin(
    adata,
    ["n_genes_by_counts", "total_counts", "pct_counts_mt"],
    jitter=0.4,
    multi_panel=True,
)
```

```python
fig, axs = plt.subplots(1, 2, figsize=(10, 4), layout="constrained")
sc.pl.scatter(adata, x="total_counts", y="pct_counts_mt", show=False, ax=axs[0])
sc.pl.scatter(adata, x="total_counts", y="n_genes_by_counts", show=False, ax=axs[1]);
```

Seuils QC adaptés à la peau (kératinocytes/fibroblastes plus riches en gènes que les PBMC) — à ajuster d'après les violins ci-dessus.

```python
adata = adata[
    (adata.obs.n_genes_by_counts > 200) & (adata.obs.n_genes_by_counts < 6000) & (adata.obs.pct_counts_mt < 15),
    :,
].copy()
adata.layers["counts"] = adata.X.copy()
adata
```

```python
sc.pp.scrublet(adata)  # détection de doublets
adata = adata[~adata.obs["predicted_doublet"]].copy()
adata
```

```python
sc.pp.normalize_total(adata, target_sum=1e4)
sc.pp.log1p(adata)
```

```python
sc.pp.highly_variable_genes(adata, layer="counts", n_top_genes=2000, flavor="seurat_v3")
sc.pl.highly_variable_genes(adata)
```

```python
adata.layers["scaled"] = adata.X.toarray()
sc.pp.regress_out(adata, ["total_counts", "pct_counts_mt"], layer="scaled")
sc.pp.scale(adata, max_value=10, layer="scaled")
```

```python
sc.pp.pca(adata, layer="scaled", svd_solver="arpack", mask_var="highly_variable")
sc.pl.pca_variance_ratio(adata, n_pcs=50, log=True)
```

```python
sc.pp.neighbors(adata, n_neighbors=15, n_pcs=30)
sc.tl.umap(adata)
sc.tl.leiden(adata, resolution=0.7, random_state=0, flavor="igraph", n_iterations=2, directed=False)
```

```python
sc.pl.umap(adata, color=["leiden", "total_counts", "pct_counts_mt"], wspace=0.4)
```

```python
marker_genes = {
    "Keratinocytes": ["KRT14", "KRT5", "KRT10", "KRT1"],
    "Fibroblasts": ["COL1A1", "DCN", "LUM"],
    "Endothelial": ["PECAM1", "VWF"],
    "Lymphatic": ["LYVE1", "CCL21"],
    "Pericytes/SMC": ["RGS5", "ACTA2"],
    "T cells": ["CD3D", "PTPRC"],
    "Myeloid": ["LYZ", "CD68", "CD1C"],
    "Mast": ["TPSAB1"],
    "Melanocytes": ["PMEL", "MLANA"],
}
marker_genes = {k: [g for g in v if g in adata.var_names] for k, v in marker_genes.items()}
sc.pl.dotplot(adata, marker_genes, groupby="leiden", standard_scale="var")
```

```python
sc.tl.rank_genes_groups(adata, "leiden", mask_var="highly_variable", method="wilcoxon")
sc.pl.rank_genes_groups(adata, n_genes=20, sharey=False)
pd.DataFrame(adata.uns["rank_genes_groups"]["names"]).head(10)
```

```python
adata.write(results_file, compression="gzip")
```
